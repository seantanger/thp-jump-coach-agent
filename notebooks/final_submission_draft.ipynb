# Vertical Jump Coach Agent: final submission (DRAFT)

**Vertical Jump Coach Agent.** A chat agent for my own dunk training, built on the THP (Train Hard Perform) jump-training transcripts. It is one chat agent with one sub-agent and an LLM judge. The chat agent (the Coach) answers questions from the transcript library with cited sources and knows my profile, my saved plan and the last few turns. When I ask for a change, it calls the Planner, a second LangGraph agent. The Planner writes the plan change, code checks it (schema, sources, calendar, spacing, equipment, per-session limits), a separate Verifier LLM judges it against the evidence, and nothing is saved until I approve it. All prompts are versioned in `jump-coach-agent/prompts/` (`coach.v5.md`, `planner.v5.md`, `verifier`). Code: https://github.com/seantanger/thp-jump-coach-agent

**I'm using it for real.** I started the plan on Nov 9 and I'm finishing week 2 this week (as of Nov 19). I haven't written week 3 yet. The saved plan holds my real logged sessions, so the demo runs against my actual training.
